```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd 
import numpy as np
```

```python
### Load the ANN train model, scalar pickle file, OneHotencoding file
model = load_model('model.h5')

## load encoder and scaler
with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender = pickle.load(file)

with open('ohe_geo.pkl', 'rb') as file:
    label_encoder_geo = pickle.load(file)


with open('scaler.pkl', 'rb') as file:
    scaler= pickle.load(file)


```

```python
## example input data
# Example input data - Test 2
input_data = {
    'CreditScore': 720,
    'Geography': 'Germany',
    'Gender': 'Female',
    'Age': 32,
    'Tenure': 7,
    'Balance': 85000,
    'NumOfProducts': 1,
    'HasCrCard': 0,
    'IsActiveMember': 0,
    'EstimatedSalary': 75000
}
```

```python
## One hot encoding "geography"
geo_encoded = label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df = pd.DataFrame(geo_encoded, columns=label_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
input_data = pd.concat([input_df.reset_index(drop=True),geo_encoded_df],axis=1)
input_data
```

```python
## Encode categoroical variable
input_df['Gender'] = label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
## concat.. with ohe data to the data
input_df=pd.concat([input_df.drop("Geography", axis=1),geo_encoded_df],axis=1)
input_df
```

```python
## scaling the input data
input_scaled = scaler.transform(input_df)
input_scaled
```

```python
## prediction 
prediction = model.predict(input_scaled)
prediction
```

```python
## prediction probab
prediction_probab = prediction[0][0]
prediction_probab
```

```python
if prediction_probab >0.5:
    print("The customer is likely to churn")
else:
    print("The customer is not likely to churn.")
```
